# SOEN Disruption Prediction — PCA-8 Sequence-Level Binary Classification

After processing all 279 SOEN steps (= 7812 raw timesteps), take the **final hidden state**
from 24 SQUID readout neurons → off-chip Linear(24→2) → softmax → binary classification.

**One prediction per subsequence**: disruptive (1) or clear (0).

**Hardware mapping**: 28 dendrite neurons × 8 synaptic inputs = 224 inputs per step.
`block_diagonal(28)`: neuron i gets PCA values from raw timestep `(step*28 + i)`.
279 RNN steps with 28×28 recurrent update. Final state → SQUID(24) → Linear(2).

**Architecture**: `Linear(224) → SingleDendrite(28, rec) → SQUID(24) → Linear(2)`
with `time_pooling: final` (extract last timestep for classification).

## 1. Convert PCA8 H5 → SOEN HDF5 format (dual-view: local + global)

Two sequential views through the same 28 neurons, concatenated into 558 steps:

**Steps 0–278 (local)**: 28 consecutive timesteps per SOEN step.
  At step s, neuron i sees PCA at raw timestep `s*28 + i`.

**Steps 279–557 (global)**: 28 equidistantly-spaced timesteps (stride=279).
  At step s, neuron i sees PCA at raw timestep `s + i*279`.
  This covers the full 7812 timesteps with dilated sampling.

The recurrent 28×28 state carries local patterns into the global pass.
Final classification from the last step of the global pass.

Training set balanced by oversampling disruptive sequences to 50/50.

```python
SOEN_H5 = DATASET_DIR / "pca8_disruption_seq2static_2class.h5"
N_NEURONS = 28
N_INPUTS = 8
INPUT_DIM = N_NEURONS * N_INPUTS  # 224


def make_local_view(X_flat):
    """(N, 7812, 8) → (N, 279, 224): 28 consecutive timesteps per step."""
    N, T, D = X_flat.shape
    T_steps = T // N_NEURONS  # 279
    X = X_flat[:, :T_steps * N_NEURONS, :]
    return X.reshape(N, T_steps, N_NEURONS * D)  # (N, 279, 224)


def make_global_view(X_flat):
    """(N, 7812, 8) → (N, 279, 224): 28 equidistant timesteps per step.

    At step s, neuron i sees raw timestep: s + i * 279.
    This gives a dilated/strided view covering the full sequence.
    """
    N, T, D = X_flat.shape
    T_steps = T // N_NEURONS  # 279 (= stride between neurons)
    X_global = np.zeros((N, T_steps, N_NEURONS * D), dtype=X_flat.dtype)
    for i in range(N_NEURONS):
        # Neuron i sees timesteps [i*279, i*279+1, ..., i*279+278]
        start = i * T_steps
        end = start + T_steps
        if end <= T:
            X_global[:, :, i*D:(i+1)*D] = X_flat[:, start:end, :]
    return X_global  # (N, 279, 224)


def oversample_minority(X, labels, seed=42):
    """Oversample disruptive sequences to match clear count → 50/50."""
    pos_idx = np.where(labels == 1)[0]
    neg_idx = np.where(labels == 0)[0]
    n_pos, n_neg = len(pos_idx), len(neg_idx)
    if n_pos == n_neg or n_pos == 0 or n_neg == 0:
        return X, labels
    rng = np.random.default_rng(seed)
    if n_pos < n_neg:
        extra = rng.choice(pos_idx, size=n_neg - n_pos, replace=True)
    else:
        extra = rng.choice(neg_idx, size=n_pos - n_neg, replace=True)
    all_idx = np.concatenate([np.arange(len(labels)), extra])
    rng.shuffle(all_idx)
    return X[all_idx], labels[all_idx]


with h5py.File(PCA8_H5, "r") as src, h5py.File(SOEN_H5, "w") as dst:
    for split in ("train", "val", "test"):
        X_raw = np.asarray(src[f"{split}/X"])       # (N, 8, 7812)
        X_flat = np.transpose(X_raw, (0, 2, 1))      # (N, 7812, 8)

        X_local = make_local_view(X_flat)             # (N, 279, 224)
        X_global = make_global_view(X_flat)            # (N, 279, 224)

        # Concatenate: local then global → (N, 558, 224)
        X = np.concatenate([X_local, X_global], axis=1)

        seq_labels = np.asarray(src[f"{split}/labels"])  # (N,)

        if split == "train":
            n_before = len(seq_labels)
            n_pos_before = int((seq_labels == 1).sum())
            X, seq_labels = oversample_minority(X, seq_labels)
            print(f"  {split}: {n_before} ({n_pos_before} disruptive) → {len(seq_labels)} (50/50)")

        g = dst.create_group(split)
        g.create_dataset("data", data=X, dtype=np.float32)
        g.create_dataset("labels", data=seq_labels, dtype=np.int64)
        g.create_dataset("input_mask", data=np.ones(X.shape[:2], dtype=bool))

        N = len(seq_labels)
        n_pos = int((seq_labels == 1).sum())
        print(f"  {split}: N={N}, data={X.shape[1:]} (279 local + 279 global = 558 steps), "
              f"disruptive={n_pos}/{N} ({n_pos/max(N,1)*100:.1f}%)")

print(f"\nSaved: {SOEN_H5}")
print(f"  Dual-view: steps 0-278 = local (consecutive), steps 279-557 = global (equidistant)")
```

## 2. Build SOEN model (block-diagonal input + SQUID readout)

`Linear(224) → SingleDendrite(28, rec) → SQUID(24) → Linear(2)`

Standard shared gamma (hardware default). Long-term dependency handled by the
**dual-view data**: local pass captures fine-grained patterns, global pass captures
sequence-wide structure, recurrent state carries context between them.

```python
MODEL_PATH = MODEL_DIR / "224IN_28H_24SQUID_2OffchipLinear_blockdiag.soen"

_build_code = f"""
from pathlib import Path
from soen_toolkit import nn

model_path = Path({str(MODEL_PATH.resolve())!r})
model_path.parent.mkdir(parents=True, exist_ok=True)

g = nn.Graph(dt=10.0, network_evaluation_method="layerwise")

# Layer 0: Input (224 = 28 neurons × 8 PCA inputs)
g.add_layer(0, nn.layers.Linear(dim=224), description="input")

# Layer 1: Hidden — 28 SingleDendrite neurons (shared gamma, hardware default)
g.add_layer(1, nn.layers.SingleDendrite(
    dim=28,
    solver="FE",
    source_func_type="RateArray",
    phi_offset=nn.param_specs.constant(0.23, learnable=False, sharing_mode="shared",
                                        constraints={{"min": -0.5, "max": 0.5}}),
    bias_current=nn.param_specs.constant(1.7, learnable=False, sharing_mode="shared",
                                          constraints={{"min": 1.1, "max": 2.0}}),
    gamma_plus=nn.param_specs.constant(2.3508e-5, learnable=False, sharing_mode="shared"),
    gamma_minus=nn.param_specs.constant(2.6995e-5, learnable=False, sharing_mode="shared",
                                         constraints={{"min": 1.7997e-5, "max": 5.3991e-5}}),
), description="hidden")

# Layer 2: SQUID readout — first 24 of 28 neurons
g.add_layer(2, nn.layers.DendriteReadout(
    dim=24,
    phi_offset=nn.param_specs.constant(0.23, learnable=False, sharing_mode="shared",
                                        constraints={{"min": -0.5, "max": 0.5}}),
    bias_current=nn.param_specs.constant(1.7, learnable=False, sharing_mode="shared",
                                          constraints={{"min": 1.1, "max": 2.0}}),
), description="readout_squid")

# Layer 3: Off-chip linear projection → 2 classes
g.add_layer(3, nn.layers.Linear(dim=2), description="offchip_projection")

# J_0_to_1: BLOCK DIAGONAL — neuron i receives only inputs [i*8:(i+1)*8]
g.connect(0, 1,
    structure=nn.structure.block_diagonal(num_heads=28),
    init=nn.init.xavier_uniform(gain=1.0),
    constraints={{"min": -0.14, "max": 0.14}},
    learnable=True,
)

# J_1_to_1: Dense recurrent (28→28)
g.connect(1, 1,
    structure=nn.structure.dense(),
    init=nn.init.orthogonal(gain=0.1),
    constraints={{"min": -0.14, "max": 0.14}},
    learnable=True,
)

# J_1_to_2: One-to-one, first 24 neurons → SQUID, fixed J=0.5
g.connect(1, 2,
    structure=nn.structure.one_to_one(source_start_node_id=0, source_end_node_id=23),
    init=nn.init.constant(0.5),
    learnable=False,
)

# J_2_to_3: Off-chip float projection (24→2), NOT quantized
g.connect(2, 3,
    structure=nn.structure.dense(),
    init=nn.init.xavier_uniform(gain=1.0),
    learnable=True,
)

g.compile()
g.save(str(model_path))
print("OK")
"""

_r = run_soen_python(_build_code, check=False)
if _r.returncode != 0:
    print("STDERR:", _r.stderr[-1000:])
    raise RuntimeError("Model build failed")

print(f"Model saved: {MODEL_PATH}")
print(f"  gamma: shared (hardware default), no per-neuron diversity")
print(f"  Long-term dependency via dual-view data (local + global passes)")
print(f"  File exists: {MODEL_PATH.exists()}")
```

## 4. Train (single run, full 200 epochs)

Run training once for all epochs. F1 evaluation done after training completes.

```python
# Single training run — no background monitor (was causing slowdowns)
import os as _os
import json as _json
import glob as _glob

print(f"Launching SOEN training: {CONFIG_PATH}")
print(f"  Python: {SOEN_PYTHON}")
print(f"  Epochs: {MAX_EPOCHS}, Batch: {BATCH_SIZE}, LR: {LEARNING_RATE}")
print(f"  Checkpoints every 50 epochs (reduces g_table overhead)")
print(f"  Metrics: accuracy, f1, precision, recall logged per epoch in SOEN output")
print()

_train_result = subprocess.run(
    [str(SOEN_PYTHON), "-m", "soen_toolkit.training", str(CONFIG_PATH.resolve())],
    env={**_os.environ, "PYTHONPATH": str(SOEN_SRC)},
)
if _train_result.returncode != 0:
    raise RuntimeError(f"Training failed with exit code {_train_result.returncode}")

print("\nTraining complete.")

# ── Final F1 evaluation on last.soen ──
def find_last_soen():
    search_roots = [str(RESULTS_DIR.resolve()), str(TUTORIAL_DIR / "training" / "results")]
    candidates = []
    for root in search_roots:
        for p in _glob.glob(f"{root}/**/last.soen", recursive=True):
            if EXPERIMENT_NAME in p:
                candidates.append(p)
    return max(candidates, key=lambda p: _os.path.getmtime(p)) if candidates else None

last_soen = find_last_soen()
if last_soen is None:
    print("WARNING: last.soen not found")
else:
    _f1_code = f"""
import sys, json, numpy as np
import h5py, torch
from soen_toolkit.core.soen_model_core import SOENModelCore

model = SOENModelCore.load({last_soen!r})
model.eval()

results = {{}}
for sp in ["train", "val", "test"]:
    with h5py.File({str(SOEN_H5.resolve())!r}, "r") as f:
        X = torch.from_numpy(np.asarray(f[sp]["data"]))
        labels = np.asarray(f[sp]["labels"])
    all_preds = []
    with torch.no_grad():
        for i in range(0, len(X), 64):
            out = model(X[i:i+64])
            if isinstance(out, tuple): out = out[0]
            all_preds.append(out[:, -1, :].argmax(dim=-1).cpu().numpy())
    preds = np.concatenate(all_preds)
    tp = int(((preds==1)&(labels==1)).sum())
    fp = int(((preds==1)&(labels==0)).sum())
    fn = int(((preds==0)&(labels==1)).sum())
    tn = int(((preds==0)&(labels==0)).sum())
    prec = tp/(tp+fp+1e-8); rec = tp/(tp+fn+1e-8)
    f1 = 2*prec*rec/(prec+rec+1e-8)
    acc = float((preds==labels).mean())
    results[sp] = {{"acc":round(acc,4),"f1":round(f1,4),"prec":round(prec,4),
                    "rec":round(rec,4),"tp":tp,"fp":fp,"fn":fn,"tn":tn}}
print(json.dumps(results))
"""
    _r = run_soen_python(_f1_code, check=False)
    if _r.returncode == 0 and _r.stdout.strip():
        try:
            f1_results = _json.loads(_r.stdout.strip().split("\n")[-1])
            print(f"\n{'='*80}")
            print(f"  FINAL F1 (last.soen)")
            print(f"{'='*80}")
            for sp in ["train", "val", "test"]:
                m = f1_results.get(sp, {})
                print(f"  {sp:>5s}: acc={m.get('acc',''):>7} f1={m.get('f1',''):>7} "
                      f"P={m.get('prec',''):>7} R={m.get('rec',''):>7} "
                      f"TP={m.get('tp','')} FP={m.get('fp','')} FN={m.get('fn','')} TN={m.get('tn','')}")
            print(f"{'='*80}")
        except _json.JSONDecodeError:
            print("F1 parse error:", _r.stdout[-300:])
    else:
        print("F1 eval failed:", _r.stderr[-500:] if _r.stderr else "")
```
